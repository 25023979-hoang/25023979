w7a16: Cho một mảng số nguyên không âm. Gọi mảng đó là "mảng đẹp" nếu
mỗi phần tử x trong mảng xuất hiện đúng x lần.
Kiểm tra xem mảng có phải là mảng đẹp hay không. Nếu không hãy xóa
đi y số (y là nhỏ nhất có thể) để mảng còn lại thành mảng đẹp.

```python
ds = list(map(int, input().split()))
ts = {}
for x in ds:
    if x in ts:
        ts[x] += 1
    else:
        ts[x] = 1

kq = 0
for x in ts:
    sl = ts[x]
    if x == 0:
        kq += sl
    elif sl >= x:
        kq += (sl - x)
    else:
        kq += sl

print(kq)
```

w7a17: Tìm phần tử xuất hiện nhiều hơn n/2 lần trong mảng (nếu có) với độ
phức tạp thời gian là O(n), độ phức tạp bộ nhớ là O(1), - Boyer–Moore.
Nếu không có phần tử nào thỏa mãn thì không in gì


```python
ds = list(map(int, input().split()))
n = len(ds)
g_tri = -1
voted = 0

for x in ds:
    if voted == 0:
        g_tri = x
        voted = 1
    elif x == g_tri:
        voted += 1
    else:
        voted -= 1

dem = 0
for x in ds:
    if x == g_tri:
        dem += 1

if dem > n // 2:
    print(g_tri)
```

w7a18: Sắp xếp có điều kiện: Số chẵn đứng trước, số lẻ đứng sau, chẵn tăng
dần, lẻ giảm dần; ; 


```python
ds = list(map(int, input().split()))
chan = []
le = []

for x in ds:
    if x % 2 == 0:
        chan.append(x)
    else:
        le.append(x)

chan.sort()
le.sort(reverse=True)

kq = chan + le
for x in kq:
    print(x, end=" ")
print()
```

w7a19: Merge 2 mảng đã sắp xếp (có thể tăng dần, có thể giảm dần) mà không
dùng bất cứ thuật toán sắp xếp nào

```python
a = list(map(int, input().split()))
b = list(map(int, input().split()))

if len(a) > 1 and a[0] > a[-1]:
    a.reverse()
if len(b) > 1 and b[0] > b[-1]:
    b.reverse()

kq = []
i = 0
j = 0

while i < len(a) and j < len(b):
    if a[i] < b[j]:
        kq.append(a[i])
        i += 1
    else:
        kq.append(b[j])
        j += 1

while i < len(a):
    kq.append(a[i])
    i += 1

while j < len(b):
    kq.append(b[j])
    j += 1

for x in kq:
    print(x, end=" ")
print()
```

w7a22: Cho N tên sinh viên gồm cả họ và tên, tìm xem tên nào hay gặp nhất.
Nếu có nhiều tên cùng số lần gặp thì lấy tên có thứ tự từ điển nhỏ nhất


```python
n = int(input())
ds = []
for i in range(n):
    ds.append(input().strip())

ts = {}
for ten in ds:
    if ten in ts:
        ts[ten] += 1
    else:
        ts[ten] = 1

max_sl = 0
for ten in ts:
    if ts[ten] > max_sl:
        max_sl = ts[ten]

kq = []
for ten in ts:
    if ts[ten] == max_sl:
        kq.append(ten)

kq.sort()
print(kq[0])
```

w7a23: Đếm số lượng tên trong xâu kí tự. Tên được định nghĩa là xâu có độ dài
>=2 và có chữ cái đầu viết hoa, các chữ cái khác viết thường, chỉ có kí

```python
chuoi = input()
tu_tach = chuoi.split()
kq = 0

for tu in tu_tach:
    if len(tu) >= 2:
        hop_le = True
        if not (tu[0] >= 'A' and tu[0] <= 'Z'):
            hop_le = False
        for i in range(1, len(tu)):
            if not (tu[i] >= 'a' and tu[i] <= 'z'):
                hop_le = False
        if hop_le:
            kq += 1

print(kq)
```

w7a25: Tìm số có đúng k chữ số lớn nhất có thể trong một số có n chữ số (coi
đầu vào là một số rất lớn - str). Không được hoán đổi vị trí các chữ số


```python
chuoi_so = input().strip()
k = int(input())
so_xoa = len(chuoi_so) - k
st = []

for s in chuoi_so:
    while so_xoa > 0 and len(st) > 0 and st[-1] < s:
        st.pop()
        so_xoa -= 1
    st.append(s)

while so_xoa > 0:
    st.pop()
    so_xoa -= 1

print("".join(st))
```

w7a26: Tách các số có trong một đoạn văn bản, tính tổng và tính trung bình các số đó.

w7a27: Cho một dãy ngoặc ()(){}{}[][]..., kiểm tra xem dãy ngoặc đã cho có hợp lệ
hay không

```python
chuoi = input().strip()
st = []
hop_le = True

for k_tu in chuoi:
    if k_tu == '(' or k_tu == '{' or k_tu == '[':
        st.append(k_tu)
    else:
        if len(st) == 0:
            hop_le = False
            break
        top = st.pop()
        if k_tu == ')' and top != '(':
            hop_le = False
            break
        if k_tu == '}' and top != '{':
            hop_le = False
            break
        if k_tu == ']' and top != '[':
            hop_le = False
            break

if len(st) > 0:
    hop_le = False

if hop_le:
    print("True")
else:
    print("False")
```

w7a28: Bà̀i toán Tìm dãy các hợp số dài nhất liên tiếp không chứa số nguyên
nào <= N (N là số nguyên dương cho trước). Nếu có nhiều dãy cùng độ
dài thì lấy dãy nhỏ nhất (gặp đầu tiên từ trái qua)


```python
def check_snt(n):
    if n < 2:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True

n = int(input())
max_len = 0
best_start = 0

curr_len = 0
curr_start = 0

for i in range(n + 1, n + 2000):
    if not check_snt(i):
        if curr_len == 0:
            curr_start = i
        curr_len += 1
    else:
        if curr_len > max_len:
            max_len = curr_len
            best_start = curr_start
        curr_len = 0

if curr_len > max_len:
    max_len = curr_len
    best_start = curr_start

for i in range(max_len):
    print(best_start + i, end=" ")
print()
```

w7a29: Tìm Dãy con tăng dần liên tục dài nhất của một danh sách nhập vào từ
bàn phím. Lấy dãy đầu tiên nếu có nhiều dãy có số phần tử bằng nhau


```python
ds = list(map(int, input().split()))

if len(ds) == 0:
    print("")
else:
    max_len = 1
    best_start = 0
    
    curr_len = 1
    curr_start = 0
    
    for i in range(1, len(ds)):
        if ds[i] > ds[i-1]:
            curr_len += 1
        else:
            if curr_len > max_len:
                max_len = curr_len
                best_start = curr_start
            curr_len = 1
            curr_start = i
            
    if curr_len > max_len:
        max_len = curr_len
        best_start = curr_start
        
    for i in range(max_len):
        print(ds[best_start + i], end=" ")
print()
```

w7a30: Tìm Dãy con tăng dần KHÔNG liên tục dài nhất của một danh sách nhập
vào từ bàn phím

```python
ds = list(map(int, input().split()))
n = len(ds)

if n == 0:
    print("")
else:
    dp = [1] * n
    tru_vet = [-1] * n
    
    for i in range(1, n):
        for j in range(i):
            if ds[i] > ds[j] and dp[i] < dp[j] + 1:
                dp[i] = dp[j] + 1
                tru_vet[i] = j
                
    max_len = 0
    best_idx = 0
    for i in range(n):
        if dp[i] > max_len:
            max_len = dp[i]
            best_idx = i
            
    kq = []
    curr = best_idx
    while curr != -1:
        kq.append(ds[curr])
        curr = tru_vet[curr]
        
    kq.reverse()
    for x in kq:
        print(x, end=" ")
print()
```
