```python
class Shape:
    def area(self):
        return 0.0

class Rectangle(Shape):
    def __init__(self, w, h):
        self.w = float(w)
        self.h = float(h)

    def area(self):
        return self.w * self.h

class Circle(Shape):
    def __init__(self, r):
        self.r = float(r)

    def area(self):
        return 3.14 * self.r * self.r

loai = input().strip()
if loai == "RECTANGLE":
    w, h = map(float, input().split())
    oo = Rectangle(w, h)
elif loai == "CIRCLE":
    r = float(input())
    oo = Circle(r)
print(f"{oo.area():.2f}")
```

```python
class Vehicle:
    def __init__(self, brand, speed):
        self.b = brand
        self.s = speed

    def info(self):
        return f"{self.b}-{self.s}"

class Car(Vehicle):
    def __init__(self, brand, speed, seats):
        Vehicle.__init__(self, brand, speed)
        self.st = seats

    def info(self):
        return f"{self.b}-{self.s}-{self.st}"

loai = input().strip()
pt = input().split()
if loai == "Vehicle":
    oo = Vehicle(pt[0], pt[1])
elif loai == "Car":
    oo = Car(pt[0], pt[1], pt[2])
print(oo.info())
```

```python
class Polygon:
    def perimeter(self):
        return 0

class Triangle(Polygon):
    def __init__(self, a, b, c):
        self.a = int(a)
        self.b = int(b)
        self.c = int(c)

    def perimeter(self):
        return self.a + self.b + self.c

a, b, c = map(int, input().split())
oo = Triangle(a, b, c)
print(oo.perimeter())
```

```python
class Person:
    def __init__(self, name, age):
        self.n = name
        self.a = age

    def describe(self):
        return f"{self.n} {self.a}"

class Student(Person):
    def __init__(self, name, age, school):
        Person.__init__(self, name, age)
        self.s = school

    def describe(self):
        return f"{self.n} {self.a} {self.s}"

loai = input().strip()
pt = input().split()
if loai == "Person":
    oo = Person(pt[0], pt[1])
elif loai == "Student":
    oo = Student(pt[0], pt[1], pt[2])
print(oo.describe())
```

```python
class Account:
    def __init__(self, balance):
        self.bal = float(balance)

    def deposit(self, x):
        self.bal += float(x)

    def get_balance(self):
        return self.bal

class SavingAccount(Account):
    def __init__(self, balance, rate):
        Account.__init__(self, balance)
        self.r = float(rate)

    def add_interest(self):
        self.bal += self.bal * self.r

b, r = map(float, input().split())
oo = SavingAccount(b, r)
oo.add_interest()
print(f"{oo.get_balance():.2f}")
```

```python
class Printer:
    def print_value(self, x):
        print(x)

class BinaryPrinter(Printer):
    def print_value(self, x):
        n = int(x)
        if n == 0:
            print(0)
            return
        kq = ""
        while n > 0:
            kq = str(n % 2) + kq
            n = n // 2
        print(kq)

v = int(input())
oo = BinaryPrinter()
oo.print_value(v)
```

```python
class Matrix2x2:
    def __init__(self, a, b, c, d):
        self.a = int(a)
        self.b = int(b)
        self.c = int(c)
        self.d = int(d)

    def trace(self):
        return self.a + self.d

    def determinant(self):
        return self.a * self.d - self.b * self.c

a, b, c, d = map(int, input().split())
oo = Matrix2x2(a, b, c, d)
print(f"{oo.trace()} {oo.determinant()}")
```

```python
class PolynomialSum:
    def __init__(self, a, b, c, d):
        self.a = int(a)
        self.b = int(b)
        self.c = int(c)
        self.d = int(d)

    def sum_str(self):
        k = self.a + self.c
        m = self.b + self.d
        chuoi = ""
        if k != 0:
            chuoi += f"{k}x"
            if m > 0:
                chuoi += f" + {m}"
            elif m < 0:
                chuoi += f" - {abs(m)}"
        else:
            chuoi += f"{m}"
        return chuoi

a, b, c, d = map(int, input().split())
oo = PolynomialSum(a, b, c, d)
print(oo.sum_str())
```

```python
class Product:
    def __init__(self, name, price):
        self.n = name
        self.p = float(price)

class DiscountProduct(Product):
    def __init__(self, name, price, percent):
        Product.__init__(self, name, price)
        self.pc = float(percent)

    def final_price(self):
        return self.p * (100.0 - self.pc) / 100.0

pt = input().split()
oo = DiscountProduct(pt[0], pt[1], pt[2])
print(f"{oo.final_price():.2f}")
```

```python
class Counter:
    def __init__(self):
        self.count = 0

    def inc(self):
        self.count += 1

    def dec(self):
        self.count -= 1

    def value(self):
        return self.count

n = int(input())
cac_lenh = input().split()
oo = Counter()

for c in cac_lenh:
    if c == "INC":
        oo.inc()
    elif c == "DEC":
        oo.dec()
        
print(oo.value())
```

```python
class Time:
    def __init__(self, h, m, s):
        self.h = int(h)
        self.m = int(m)
        self.s = int(s)

    def to_seconds(self):
        return self.h * 3600 + self.m * 60 + self.s

h, m, s = map(int, input().split())
oo = Time(h, m, s)
print(oo.to_seconds())
```

```python
class Stack:
    def __init__(self):
        self.st = []

    def push(self, x):
        self.st.append(x)

    def pop(self):
        if len(self.st) > 0:
            self.st.pop()

    def top(self):
        if len(self.st) > 0:
            return self.st[-1]
        return None

n = int(input())
oo = Stack()

for i in range(n):
    dong = input().split()
    cmd = dong[0]
    
    if cmd == "PUSH":
        oo.push(dong[1])
    elif cmd == "POP":
        oo.pop()
    elif cmd == "TOP":
        print(oo.top())
```
