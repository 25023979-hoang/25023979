Bài 2B: Viết hàm zigzag_numbers(n) trả về một chuỗi biểu diễn ma trận n x n chứa các số từ 1 đến n^2. Các hàng cách nhau bởi ký tự xuống 
dòng. Với chỉ số hàng bắt đầu từ 0: hàng chẵn đi từ trái sang phải, hàng lẻ đi từ phải sang trái. Các số trong cùng một hàng cách nhau bởi 
một dấu cách. 
n = 2
1 2
4 3
n= 4
1 2 3 4
8 7 6 5
9 10 11 12
16 15 14 13

```python
def zigzag_numbers(n):
    mt = []
    x = 1
    for i in range(n):
        hang = []
        for j in range(n):
            hang.append(x)
            x = x + 1
        if i % 2 == 1:
            hang.reverse()
        mt.append(hang)
        
    chuoi_kq = ""
    for i in range(n):
        dong_chu = ""
        for j in range(n):
            dong_chu += str(mt[i][j]) + " "
        chuoi_kq += dong_chu.strip() + "\n"
        
    return chuoi_kq.strip()

n = int(input())
print(zigzag_numbers(n))
```

Bài 4: Trả về hàng (chỉ số bắt đầu từ 0) có tổng hàng nhỏ nhất trong ma trận n x m. 
Viết hàm row_with_min_sum(mat) trả về chỉ số bắt đầu từ 0 của hàng có tổng nhỏ nhất. 
Nếu có nhiều hàng cùng tổng nhỏ nhất, trả về hàng có chỉ số lớn nhất. - Dùng vòng lặp để tính tổng từng hàng. -. 
n = 2
m = 3
mat = [[1, 3, 5], 
       [6, 2, -9]]


```python
def row_with_min_sum(mat):
    sum_min = 1000000000
    row_min = -1
    n = len(mat)
    m = len(mat[0])
    for i in range(n):
        s = 0
        for j in range(m):
            s += mat[i][j]
        if s <= sum_min:
            sum_min = s
            row_min = i
    return row_min, sum_min
n, m = map(int, input().split())
matrix = []
for i in range(n):
    row = list(map(int, input().split()))
    matrix.append(row)
print(row_with_min_sum(matrix))
```

Bài 5: 
Cho ma trận vuông cấp n. In ra đường chéo chính và đường chéo phụ của ma trận


```python
n,m = list(map(int,input().split()))
mt = []
for i in range (n):
    hang = list(map(int,input().split()))
    mt.append(hang)
dcc=[]
for i in range(n):
    for j in range(m):
        if i == j :
            dcc.append(mt[i][j])
print(dcc)
dcp = []
for i in range(n):
    dcp.append(mt[i][n - 1 - i])
print(dcp)
```

Bài 6: Cho ma trận nxm số nguyên, thêm vào một cột tại vị trí k.

```python
n,m,k = list(map(int,input().split()))
mt = []
for i in range (n):
    hang = list(map(int,input().split()))
    mt.append(hang)
cotk = list(map(int,input().split()))
for i in range(n):
    mt[i].insert(k, cotk[i])
for row in mt:
    for x in row:
        print(x,end=' ')
    print()
```
